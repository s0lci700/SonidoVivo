```python
import pandas as pd
import unicodedata
def remove_tildes(text):
    # Normalize string to decompose characters (e.g., 'á' becomes 'a' + '◌́')
    normalized = unicodedata.normalize('NFD', text)
    # Filter out characters that are diacritic marks ('Mn' category)
    return ''.join(c for c in normalized if unicodedata.category(c) != 'Mn')


DF = pd.read_excel('DSY1104 - Forma B - Catalogo Sonido Vivo.xlsx', )
headers = DF[1:2].values.tolist()[0]
headers = [str(header).split(' ')[0].lower() for header in headers]
headers = [remove_tildes(header) for header in headers]
headers
```

```python
content = DF[2:].values.tolist()
content
```

```python
output = [dict(zip(headers, row)) for row in content]
output
```

```python
for p in output:
    p['categoriaId'] = p['codigo'][:2]
output
```
